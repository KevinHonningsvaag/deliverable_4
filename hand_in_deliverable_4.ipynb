### Creating and simplifying fractions

The constructor validates the numerator and denominator before
creating the Fraction object. Both values must be integers,
and the denominator must be above zero and positive.

The greatest common divisor (GCD) is used to automatically
simplify the fraction to its lowest it can be.
to show how we really view fractions in the real worl

```python
#Example
fraction1 = Fraction(13, 18) #THis fraction cant be simplified
fraction2 = Fraction(-6, 8)
fraction3 = Fraction(0, 5)

print(f"13/18 = {fraction1.numerator}/{fraction1.denominator}")
print(f"-6/8 = {fraction2.numerator}/{fraction2.denominator}")
print(f"0/5 = {fraction3.numerator}/{fraction3.denominator}")
```

## Different dunder methods

```python
a = Fraction(1, 2)
b = Fraction(2, 4)
c = Fraction(3, 4)

print("Equal (__eq__):", a == b)
print("Not equal (__ne__):", a != c)

print("Less than (__lt__):", a < c)
print("Less than or equal (__le__):", a <= b)

print("Greater than (__gt__):", c > a)
print("Greater than or equal (__ge__):", a >= b)

```


## Dunder/arithmetic methods of own choosing

Here we had to choose wich four methods to incorporate ourselves, which led us to a page about dundermethods. Here we saw that addition, subtraction, multiplication and division looked the most relevant for this project an therefore picked those four.

```python
a = Fraction(1, 2)
b = Fraction(2, 4)
c = Fraction(3, 4)

def show_fraction(fraction: Fraction) -> str:
    """Return a fraction as a readable string."""
    return f"{fraction.numerator}/{fraction.denominator}"


print("Addition (__add__):", show_fraction(a + c))
print("Subtraction (__sub__):", show_fraction(a - c))
print("Multiplication (__mul__):", show_fraction(a * c))
print("Division (__truediv__):", show_fraction(a / c))
print("Negative division:", show_fraction(Fraction(2, 3) / Fraction(-4, 5)))
```

Here we can see that the script also handles negative fractions, so that the denominator will not become negative. This is handled by multiplying the new numerator and denominator by -1 if the denominator becomes negative during the truediv operaton.

## Error handling for task 2

The class validates the input values to prevent invalid mathematical operations, such as division by 0

```python
try:
    Fraction(1, 0)
except ValueError as error:
    print("Invalid denominator:", error)

try:
    Fraction(1, -1)
except ValueError as error:
    print("Negative denominator:", error)

try:
    Fraction(1.5, 2)
except TypeError as error:
    print("Invalid type:", error)

try:
    Fraction(1, 2) / Fraction(0, 1)
except ZeroDivisionError:
    print("Division by zero is not allowed")

```
